## 1. Target Distribution Analysis

Verify that vulnerable and patched samples are balanced

````python
# Target distribution analysis
target_counts = df_paired['target'].value_counts().sort_index()
total_samples = len(df_paired)

output = StringIO()
output.write("## 1. Target Distribution Analysis\n\n")
output.write("### Overview\n")
output.write(f"- **Total samples:** {total_samples}\n")
output.write(f"- **Vulnerable (target=1):** {target_counts.get(1, 0)} ({target_counts.get(1, 0)/total_samples*100:.2f}%)\n")
output.write(f"- **Patched (target=0):** {target_counts.get(0, 0)} ({target_counts.get(0, 0)/total_samples*100:.2f}%)\n\n")

# Check balance
balance_ratio = target_counts.get(1, 0) / target_counts.get(0, 0) if target_counts.get(0, 0) > 0 else 0
output.write(f"### Balance Ratio\n")
output.write(f"- **Vulnerable:Patched ratio:** {balance_ratio:.3f}\n")
if abs(balance_ratio - 1.0) < 0.1:
    output.write(f"- **Status:** ✅ Well-balanced (within 10% of 1:1)\n\n")
elif abs(balance_ratio - 1.0) < 0.3:
    output.write(f"- **Status:** ⚠️ Moderately imbalanced\n\n")
else:
    output.write(f"- **Status:** ❌ Highly imbalanced - investigate pairing logic\n\n")

# Visualization
plt.figure(figsize=(8, 6))
target_counts.plot(kind='bar', color=['#2ecc71', '#e74c3c'])
plt.title('Distribution of Vulnerable vs Patched Samples')
plt.xlabel('Target (0=Patched, 1=Vulnerable)')
plt.ylabel('Count')
plt.xticks([0, 1], ['Patched (0)', 'Vulnerable (1)'], rotation=0)
plt.tight_layout()
target_plot_path = output_dir / "train_paired" / "target_distribution.png"
plt.savefig(target_plot_path)
plt.close()

output.write(f"![Target Distribution]({target_plot_path.name})\n\n")

# Write to report
with open(report_path, "a") as f:
    f.write(output.getvalue())

print(output.getvalue())
````

## 2. Pairing Validation

Check if consecutive rows form proper vulnerable/patched pairs

````python
# Pairing validation - check if rows form proper pairs
output = StringIO()
output.write("## 2. Pairing Validation\n\n")

# Group by CVE and check pairing
cve_groups = df_paired.groupby('cve')

# Statistics
cves_with_both = 0
cves_only_vuln = 0
cves_only_patch = 0
cves_with_multiple = 0
perfect_pairs = 0

pairing_details = []

for cve, group in cve_groups:
    vuln_count = (group['target'] == 1).sum()
    patch_count = (group['target'] == 0).sum()
    
    if vuln_count > 0 and patch_count > 0:
        cves_with_both += 1
        if vuln_count == 1 and patch_count == 1:
            perfect_pairs += 1
        elif vuln_count > 1 or patch_count > 1:
            cves_with_multiple += 1
            pairing_details.append({
                'cve': cve,
                'vuln_count': vuln_count,
                'patch_count': patch_count,
                'total': len(group)
            })
    elif vuln_count > 0:
        cves_only_vuln += 1
    elif patch_count > 0:
        cves_only_patch += 1

total_cves = len(cve_groups)

output.write(f"### Pairing Statistics\n")
output.write(f"- **Total CVEs:** {total_cves}\n")
output.write(f"- **Perfect pairs (1 vuln + 1 patch):** {perfect_pairs} ({perfect_pairs/total_cves*100:.2f}%)\n")
output.write(f"- **CVEs with multiple versions:** {cves_with_multiple} ({cves_with_multiple/total_cves*100:.2f}%)\n")
output.write(f"- **CVEs with only vulnerable code:** {cves_only_vuln} ({cves_only_vuln/total_cves*100:.2f}%)\n")
output.write(f"- **CVEs with only patched code:** {cves_only_patch} ({cves_only_patch/total_cves*100:.2f}%)\n\n")

if cves_only_vuln > 0 or cves_only_patch > 0:
    output.write(f"⚠️ **Warning:** {cves_only_vuln + cves_only_patch} CVEs have incomplete pairs!\n\n")
else:
    output.write(f"✅ **All CVEs have both vulnerable and patched versions**\n\n")

# Show examples of CVEs with multiple versions
if pairing_details:
    output.write(f"### CVEs with Multiple Versions (Top 10)\n\n")
    pairing_df = pd.DataFrame(pairing_details).sort_values('total', ascending=False).head(10)
    output.write(pairing_df.to_markdown(index=False))
    output.write("\n\n")

# Check consecutive pairing pattern
consecutive_pairs = 0
for i in range(0, len(df_paired)-1, 2):
    if i+1 < len(df_paired):
        row1 = df_paired.iloc[i]
        row2 = df_paired.iloc[i+1]
        if (row1['cve'] == row2['cve'] and 
            row1['target'] != row2['target']):
            consecutive_pairs += 1

expected_consecutive = total_samples // 2
output.write(f"### Consecutive Pairing Pattern\n")
output.write(f"- **Consecutive pairs found:** {consecutive_pairs}\n")
output.write(f"- **Expected (total/2):** {expected_consecutive}\n")
output.write(f"- **Match rate:** {consecutive_pairs/expected_consecutive*100:.2f}%\n\n")

if consecutive_pairs / expected_consecutive > 0.9:
    output.write("✅ **Dataset follows consecutive pairing pattern**\n\n")
else:
    output.write("⚠️ **Dataset does NOT follow strict consecutive pairing**\n\n")

# Write to report
with open(report_path, "a") as f:
    f.write(output.getvalue())

print(output.getvalue())
````

## 3. CWE Distribution Analysis

Analyze vulnerability types in the dataset

````python
# CWE Distribution Analysis
import ast

output = StringIO()
output.write("## 3. CWE Distribution Analysis\n\n")

# Parse CWE field (it's a list stored as string)
all_cwes = []
multi_cwe_count = 0

for cwe_val in df_paired['cwe']:
    if isinstance(cwe_val, str):
        try:
            cwe_list = ast.literal_eval(cwe_val)
            if isinstance(cwe_list, list):
                all_cwes.extend(cwe_list)
                if len(cwe_list) > 1:
                    multi_cwe_count += 1
            else:
                all_cwes.append(cwe_list)
        except:
            all_cwes.append(cwe_val)
    elif isinstance(cwe_val, list):
        all_cwes.extend(cwe_val)
        if len(cwe_val) > 1:
            multi_cwe_count += 1

cwe_counts = pd.Series(all_cwes).value_counts()

output.write(f"### Overview\n")
output.write(f"- **Total CWE tags:** {len(all_cwes)}\n")
output.write(f"- **Unique CWE types:** {len(cwe_counts)}\n")
output.write(f"- **Samples with multiple CWEs:** {multi_cwe_count} ({multi_cwe_count/total_samples*100:.2f}%)\n")
output.write(f"- **Avg CWEs per sample:** {len(all_cwes)/total_samples:.2f}\n\n")

# Top CWE categories
output.write(f"### Top 20 CWE Categories\n\n")
top_cwes = cwe_counts.head(20)
cwe_df = pd.DataFrame({
    'CWE': top_cwes.index,
    'Count': top_cwes.values,
    'Percentage': (top_cwes.values / len(all_cwes) * 100).round(2)
})
output.write(cwe_df.to_markdown(index=False))
output.write("\n\n")

# Visualization
plt.figure(figsize=(14, 8))
top_cwes.head(15).plot(kind='barh', color='#3498db')
plt.title('Top 15 CWE Categories')
plt.xlabel('Number of Samples')
plt.ylabel('CWE')
plt.gca().invert_yaxis()
plt.tight_layout()
cwe_plot_path = output_dir / "train_paired" / "cwe_distribution.png"
plt.savefig(cwe_plot_path)
plt.close()

output.write(f"![CWE Distribution]({cwe_plot_path.name})\n\n")

# CWE category mapping (common ones)
cwe_categories = {
    'CWE-119': 'Buffer Overflow',
    'CWE-120': 'Buffer Copy without Checking Size',
    'CWE-125': 'Out-of-bounds Read',
    'CWE-787': 'Out-of-bounds Write',
    'CWE-416': 'Use After Free',
    'CWE-476': 'NULL Pointer Dereference',
    'CWE-190': 'Integer Overflow',
    'CWE-189': 'Numeric Errors',
    'CWE-200': 'Information Exposure',
    'CWE-20': 'Improper Input Validation',
    'CWE-79': 'Cross-site Scripting (XSS)',
    'CWE-89': 'SQL Injection',
    'CWE-264': 'Permissions/Privileges',
    'CWE-310': 'Cryptographic Issues',
    'CWE-362': 'Race Condition',
}

output.write(f"### Common CWE Categories Found\n\n")
for cwe, desc in cwe_categories.items():
    if cwe in cwe_counts.index:
        count = cwe_counts[cwe]
        output.write(f"- **{cwe}** ({desc}): {count} samples\n")
output.write("\n")

# Write to report
with open(report_path, "a") as f:
    f.write(output.getvalue())

print(output.getvalue())
````

## 4. Code Length Analysis

Compare function sizes between vulnerable and patched versions

````python
# Code Length Analysis
output = StringIO()
output.write("## 4. Code Length Analysis\n\n")

# Calculate LOC for each function
df_paired['loc'] = df_paired['func'].apply(lambda x: len([l for l in str(x).split('\n') if l.strip()]))

# Separate by target
vuln_loc = df_paired[df_paired['target'] == 1]['loc']
patch_loc = df_paired[df_paired['target'] == 0]['loc']

output.write(f"### Lines of Code Statistics\n\n")
output.write(f"**Vulnerable Functions:**\n")
output.write(f"- Mean: {vuln_loc.mean():.2f} lines\n")
output.write(f"- Median: {vuln_loc.median():.2f} lines\n")
output.write(f"- Std Dev: {vuln_loc.std():.2f}\n")
output.write(f"- Min: {vuln_loc.min()} lines\n")
output.write(f"- Max: {vuln_loc.max()} lines\n\n")

output.write(f"**Patched Functions:**\n")
output.write(f"- Mean: {patch_loc.mean():.2f} lines\n")
output.write(f"- Median: {patch_loc.median():.2f} lines\n")
output.write(f"- Std Dev: {patch_loc.std():.2f}\n")
output.write(f"- Min: {patch_loc.min()} lines\n")
output.write(f"- Max: {patch_loc.max()} lines\n\n")

# Calculate difference for paired samples
# Group by CVE and calculate LOC delta
cve_deltas = []
for cve, group in df_paired.groupby('cve'):
    vuln = group[group['target'] == 1]
    patch = group[group['target'] == 0]
    if len(vuln) > 0 and len(patch) > 0:
        # Take first of each if multiple
        delta = patch.iloc[0]['loc'] - vuln.iloc[0]['loc']
        cve_deltas.append(delta)

if cve_deltas:
    cve_deltas_series = pd.Series(cve_deltas)
    output.write(f"### LOC Delta (Patched - Vulnerable)\n\n")
    output.write(f"- **Mean delta:** {cve_deltas_series.mean():.2f} lines\n")
    output.write(f"- **Median delta:** {cve_deltas_series.median():.2f} lines\n")
    output.write(f"- **Patches that add code:** {(cve_deltas_series > 0).sum()} ({(cve_deltas_series > 0).sum()/len(cve_deltas)*100:.2f}%)\n")
    output.write(f"- **Patches that remove code:** {(cve_deltas_series < 0).sum()} ({(cve_deltas_series < 0).sum()/len(cve_deltas)*100:.2f}%)\n")
    output.write(f"- **Patches with same LOC:** {(cve_deltas_series == 0).sum()} ({(cve_deltas_series == 0).sum()/len(cve_deltas)*100:.2f}%)\n\n")

# Visualization - LOC distribution
fig, axes = plt.subplots(1, 2, figsize=(20, 8))  # Larger figure

# Use kernel density estimation for smoother comparison
vuln_loc.plot.kde(ax=axes[0], color='#e74c3c', linewidth=2, label='Vulnerable')
patch_loc.plot.kde(ax=axes[0], color='#2ecc71', linewidth=2, label='Patched')
axes[0].set_xlabel('Lines of Code')
axes[0].set_ylabel('Density')
axes[0].set_title('LOC Distribution: Vulnerable vs Patched')
axes[0].legend()
axes[0].set_xlim(0, 200)

# Delta histogram
if cve_deltas:
    axes[1].hist(cve_deltas_series, bins=50, color='#3498db', alpha=0.7)
    axes[1].axvline(0, color='red', linestyle='--', linewidth=2, label='No change')
    axes[1].set_xlabel('LOC Delta (Patched - Vulnerable)')
    axes[1].set_ylabel('Frequency')
    axes[1].set_title('Distribution of LOC Changes')
    axes[1].legend()
    axes[1].set_xlim(-50, 50)  # Focus on main distribution

plt.tight_layout()
loc_plot_path = output_dir / "train_paired" / "loc_analysis.png"
plt.savefig(loc_plot_path)
plt.close()

output.write(f"![LOC Analysis]({loc_plot_path.name})\n\n")

# Write to report
with open(report_path, "a") as f:
    f.write(output.getvalue())

print(output.getvalue())
````

## 5. CWE by Project Analysis

Identify project-specific vulnerability patterns

````python
# CWE by Project Analysis
output = StringIO()
output.write("## 5. CWE by Project Analysis\n\n")

# Get top projects
top_projects_list = df_paired['project'].value_counts().head(10).index.tolist()

# Extract primary CWE for each sample
def get_primary_cwe(cwe_val):
    if isinstance(cwe_val, str):
        try:
            cwe_list = ast.literal_eval(cwe_val)
            if isinstance(cwe_list, list) and len(cwe_list) > 0:
                return cwe_list[0]
            return cwe_list
        except:
            return cwe_val
    elif isinstance(cwe_val, list) and len(cwe_val) > 0:
        return cwe_val[0]
    return cwe_val

df_paired['primary_cwe'] = df_paired['cwe'].apply(get_primary_cwe)

# Create crosstab for top projects and top CWEs
top_cwes_list = df_paired['primary_cwe'].value_counts().head(10).index.tolist()

# Filter data
df_filtered = df_paired[
    (df_paired['project'].isin(top_projects_list)) &
    (df_paired['primary_cwe'].isin(top_cwes_list))
]

# Create crosstab
crosstab = pd.crosstab(df_filtered['project'], df_filtered['primary_cwe'])

output.write(f"### Top 10 Projects vs Top 10 CWEs\n\n")
output.write("Crosstab showing vulnerability type distribution across major projects:\n\n")
output.write(crosstab.to_markdown())
output.write("\n\n")

# Find project-CWE associations
project_cwe_analysis = []
for project in top_projects_list:
    project_data = df_paired[df_paired['project'] == project]
    if len(project_data) > 0:
        cwe_dist = project_data['primary_cwe'].value_counts().head(3)
        top_cwe = cwe_dist.index[0] if len(cwe_dist) > 0 else 'N/A'
        top_cwe_pct = (cwe_dist.values[0] / len(project_data) * 100) if len(cwe_dist) > 0 else 0
        
        project_cwe_analysis.append({
            'Project': project,
            'Total_Samples': len(project_data),
            'Top_CWE': top_cwe,
            'Top_CWE_Count': cwe_dist.values[0] if len(cwe_dist) > 0 else 0,
            'Top_CWE_Pct': f"{top_cwe_pct:.1f}%"
        })

output.write(f"### Dominant CWE per Project\n\n")
project_cwe_df = pd.DataFrame(project_cwe_analysis)
output.write(project_cwe_df.to_markdown(index=False))
output.write("\n\n")

# Heatmap visualization
plt.figure(figsize=(14, 10))
import seaborn as sns
sns.heatmap(crosstab, annot=True, fmt='d', cmap='YlOrRd', cbar_kws={'label': 'Sample Count'})
plt.title('Vulnerability Types (CWE) by Project')
plt.xlabel('CWE Category')
plt.ylabel('Project')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
heatmap_path = output_dir / "train_paired" / "cwe_by_project_heatmap.png"
plt.savefig(heatmap_path)
plt.close()

output.write(f"![CWE by Project Heatmap]({heatmap_path.name})\n\n")

# Write to report
with open(report_path, "a") as f:
    f.write(output.getvalue())

print(output.getvalue())
````

## 6. Code Pattern Analysis

Search for common vulnerability and fix patterns

````python
# Code Pattern Analysis
import re

output = StringIO()
output.write("## 6. Code Pattern Analysis\n\n")

# Define patterns to search for
vuln_patterns = {
    'unsafe_strcpy': r'\bstrcpy\s*\(',
    'unsafe_strcat': r'\bstrcat\s*\(',
    'unsafe_sprintf': r'\bsprintf\s*\(',
    'unsafe_gets': r'\bgets\s*\(',
    'malloc_call': r'\b(malloc|calloc|realloc)\s*\(',
    'free_call': r'\bfree\s*\(',
    'memcpy_call': r'\bmemcpy\s*\(',
    'pointer_deref': r'->',
    'array_access': r'\[.*\]',
}

patch_patterns = {
    'safe_strncpy': r'\bstrncpy\s*\(',
    'safe_strncat': r'\bstrncat\s*\(',
    'safe_snprintf': r'\bsnprintf\s*\(',
    'null_check': r'if\s*\([^)]*!=\s*NULL\)',
    'size_check': r'if\s*\([^)]*(<|>|<=|>=)[^)]*\)',
    'bounds_check': r'if\s*\([^)]*(size|len|length|count)[^)]*\)',
    'return_check': r'if\s*\([^)]*<\s*0\)',
}

# Count patterns in vulnerable vs patched code
vuln_pattern_counts = {pattern: 0 for pattern in vuln_patterns}
patch_pattern_counts = {pattern: 0 for pattern in patch_patterns}

for _, row in df_paired.iterrows():
    code = str(row['func'])
    is_vuln = row['target'] == 1
    
    if is_vuln:
        for pattern_name, regex in vuln_patterns.items():
            if re.search(regex, code, re.IGNORECASE):
                vuln_pattern_counts[pattern_name] += 1
    else:
        for pattern_name, regex in patch_patterns.items():
            if re.search(regex, code, re.IGNORECASE):
                patch_pattern_counts[pattern_name] += 1

output.write(f"### Vulnerable Code Patterns\n\n")
output.write(f"Patterns found in vulnerable functions:\n\n")
vuln_df = pd.DataFrame({
    'Pattern': list(vuln_pattern_counts.keys()),
    'Count': list(vuln_pattern_counts.values()),
    'Percentage': [f"{(v/len(df_paired[df_paired['target']==1])*100):.2f}%" for v in vuln_pattern_counts.values()]
}).sort_values('Count', ascending=False)
output.write(vuln_df.to_markdown(index=False))
output.write("\n\n")

output.write(f"### Patch/Fix Patterns\n\n")
output.write(f"Patterns found in patched functions:\n\n")
patch_df = pd.DataFrame({
    'Pattern': list(patch_pattern_counts.keys()),
    'Count': list(patch_pattern_counts.values()),
    'Percentage': [f"{(v/len(df_paired[df_paired['target']==0])*100):.2f}%" for v in patch_pattern_counts.values()]
}).sort_values('Count', ascending=False)
output.write(patch_df.to_markdown(index=False))
output.write("\n\n")

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Vulnerable patterns
vuln_df.sort_values('Count').tail(10).plot(x='Pattern', y='Count', kind='barh', ax=axes[0], color='#e74c3c', legend=False)
axes[0].set_title('Top 10 Vulnerable Code Patterns')
axes[0].set_xlabel('Count')
axes[0].set_ylabel('')

# Patch patterns
patch_df.sort_values('Count').tail(10).plot(x='Pattern', y='Count', kind='barh', ax=axes[1], color='#2ecc71', legend=False)
axes[1].set_title('Top 10 Patch/Fix Patterns')
axes[1].set_xlabel('Count')
axes[1].set_ylabel('')

plt.tight_layout()
pattern_plot_path = output_dir / "train_paired" / "code_patterns.png"
plt.savefig(pattern_plot_path)
plt.close()

output.write(f"![Code Patterns]({pattern_plot_path.name})\n\n")

# Write to report
with open(report_path, "a") as f:
    f.write(output.getvalue())

print(output.getvalue())
````

## 7. Function Completeness Check

Verify if functions are complete for Coming analysis

````python
# Function Completeness Check
output = StringIO()
output.write("## 7. Function Completeness Analysis\n\n")

def check_completeness(code):
    """Check if function appears complete"""
    code_str = str(code)
    
    # Check for opening and closing braces
    open_braces = code_str.count('{')
    close_braces = code_str.count('}')
    
    # Check for function signature
    has_signature = bool(re.search(r'\w+\s+\w+\s*\([^)]*\)', code_str))
    
    # Check if braces are balanced
    balanced_braces = open_braces == close_braces and open_braces > 0
    
    return {
        'has_signature': has_signature,
        'balanced_braces': balanced_braces,
        'open_braces': open_braces,
        'close_braces': close_braces,
        'complete': has_signature and balanced_braces
    }

# Analyze completeness
completeness_results = df_paired['func'].apply(check_completeness)
df_paired['is_complete'] = completeness_results.apply(lambda x: x['complete'])

complete_count = df_paired['is_complete'].sum()
incomplete_count = len(df_paired) - complete_count

output.write(f"### Completeness Statistics\n\n")
output.write(f"- **Complete functions:** {complete_count} ({complete_count/len(df_paired)*100:.2f}%)\n")
output.write(f"- **Incomplete/snippets:** {incomplete_count} ({incomplete_count/len(df_paired)*100:.2f}%)\n\n")

# Check by target
vuln_complete = df_paired[df_paired['target'] == 1]['is_complete'].sum()
patch_complete = df_paired[df_paired['target'] == 0]['is_complete'].sum()
vuln_total = len(df_paired[df_paired['target'] == 1])
patch_total = len(df_paired[df_paired['target'] == 0])

output.write(f"### Completeness by Target\n\n")
output.write(f"**Vulnerable functions:**\n")
output.write(f"- Complete: {vuln_complete}/{vuln_total} ({vuln_complete/vuln_total*100:.2f}%)\n\n")

output.write(f"**Patched functions:**\n")
output.write(f"- Complete: {patch_complete}/{patch_total} ({patch_complete/patch_total*100:.2f}%)\n\n")

# Check paired completeness
paired_complete = 0
for cve, group in df_paired.groupby('cve'):
    vuln = group[group['target'] == 1]
    patch = group[group['target'] == 0]
    if len(vuln) > 0 and len(patch) > 0:
        if vuln.iloc[0]['is_complete'] and patch.iloc[0]['is_complete']:
            paired_complete += 1

total_cves = df_paired['cve'].nunique()
output.write(f"### Paired Completeness\n\n")
output.write(f"- **CVEs with both functions complete:** {paired_complete}/{total_cves} ({paired_complete/total_cves*100:.2f}%)\n")
output.write(f"- **Suitable for Coming analysis:** {paired_complete} pairs\n\n")

if paired_complete < total_cves * 0.5:
    output.write(f"⚠️ **Warning:** Less than 50% of pairs have complete functions. May need to use file_contents/ for full context.\n\n")
else:
    output.write(f"✅ **Most pairs have complete functions, suitable for Coming analysis.**\n\n")

# Sample incomplete functions
incomplete_samples = df_paired[~df_paired['is_complete']].head(3)
if len(incomplete_samples) > 0:
    output.write(f"### Examples of Incomplete Functions\n\n")
    for idx, row in incomplete_samples.iterrows():
        output.write(f"**CVE:** {row['cve']}, **Target:** {row['target']}\n")
        func_preview = str(row['func'])[:200].replace('\n', ' ')
        output.write(f"```\n{func_preview}...\n```\n\n")

# Write to report
with open(report_path, "a") as f:
    f.write(output.getvalue())

print(output.getvalue())
````
