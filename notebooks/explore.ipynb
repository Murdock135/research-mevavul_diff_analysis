```python
import json
from pathlib import Path

import pandas as pd
```

```python
# Load the validation paired data (most manageable size)
path = Path("../PrimeVul_v0.1/primevul_valid_paired.jsonl")

# Read JSONL file (line-delimited JSON)
data = []
with open(path, 'r') as f:
    for line in f:
        data.append(json.loads(line))

df = pd.DataFrame(data)
print(f"Loaded {len(df)} entries")
print(f"Shape: {df.shape}")
df.head()
```

```python
# Check target distribution (should be 50/50 for paired data)
print("Target distribution:")
print(df['target'].value_counts())
print(f"\nTotal pairs: {len(df) // 2}")
```

```python
# Check column names and data types
print("Columns:", df.columns.tolist())
print("\nData types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Load file_info.json (it's large, so be careful)
file_info_path = Path("../PrimeVul_v0.1/file_info.json")

# Check file size first
import os
file_size_mb = os.path.getsize(file_info_path) / (1024 * 1024)
print(f"file_info.json size: {file_size_mb:.2f} MB")

# Try to load it
with open(file_info_path, 'r') as f:
    file_info = json.load(f)

print(f"Type: {type(file_info)}")
if isinstance(file_info, dict):
    print(f"Number of entries: {len(file_info)}")
    # Show first few keys
    sample_keys = list(file_info.keys())[:5]
    print(f"\nSample keys: {sample_keys}")
    # Show structure of first entry
    if sample_keys:
        print(f"\nStructure of first entry:")
        print(json.dumps({sample_keys[0]: file_info[sample_keys[0]]}, indent=2))
elif isinstance(file_info, list):
    print(f"Number of entries: {len(file_info)}")
    print(f"\nFirst entry:")
    print(json.dumps(file_info[0], indent=2))
```
