```python
import json
from pathlib import Path

import pandas as pd
```

```python
results_dir = Path("./results")
```

# Basic Info on file_info.json

```python
# Load the validation paired data (most manageable size)
path = Path("../PrimeVul_v0.1/file_info.json")

with open(path, "r") as f:
    data = json.load(f)

df = pd.DataFrame(data)

print("================================= file_info.json =============================")
print("--------------------------------- DataFrame info ---------------------------------")
print(df.info())
print(f"Samples: {len(df)}")

df.head()

# Write head of file_info.json into file
with open(f"{results_dir}/file_info_head.json", "w") as f:
    json.dump(df.head().to_dict(), f, indent=4)
```

```python
# Transpose file_info
df_transposed = df.T
print("--------------------------------- Transposed DataFrame info ---------------------------------")
print(df_transposed.info())
print("A few samples")
df_transposed.head()

# Write head into a file
with open(f"{results_dir}/file_info_t_head.json", "w") as f:
    json.dump(df_transposed.head().to_dict(), f, indent=4)
```
