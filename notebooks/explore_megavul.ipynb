```python
from pprint import pformat

def preview_value(value, max_str=220, max_items=3):
    if isinstance(value, str):
        compact = value.replace("\n", "\\n")
        return compact[:max_str] + ("..." if len(compact) > max_str else "")
    if isinstance(value, list):
        head = value[:max_items]
        return f"len={len(value)} | head={pformat(head, compact=True)}"
    if isinstance(value, dict):
        keys = list(value.keys())
        head_keys = keys[:max_items]
        head = {k: value[k] for k in head_keys}
        return f"keys={keys[:10]}{'...' if len(keys) > 10 else ''} | sample={pformat(head, compact=True)}"
    return repr(value)

def demo_entry_samples(entry, title):
    print(f"\n=== {title} ===")
    for key, value in entry.items():
        print(f"\n{key} ({type(value).__name__})")
        print(preview_value(value))

vuln_entry = vulnerable_functions[0]
non_vuln_entry = non_vulnerable_functions[0]

demo_entry_samples(vuln_entry, "vulnerable_functions[0] field samples")
demo_entry_samples(non_vuln_entry, "non_vulnerable_functions[0] field samples")
```
