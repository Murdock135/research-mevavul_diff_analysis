```python
# Quick diagnostic script
import pandas as pd

features_df = pd.read_csv('../data/gumtree_features.csv')
pairs_df = pd.read_csv('../data/code_pairs.csv')

# Check the xml_path format
print("Sample xml_paths:")
print(features_df['xml_path'].head(10))

# Check the id range in code_pairs
print("\nCode pairs id range:")
print(f"Min: {pairs_df['id'].min()}, Max: {pairs_df['id'].max()}")

# Extract pair_id and check
features_df['pair_id'] = features_df['xml_path'].str.extract(r'pair_(\d+)/').astype(int)
print("\nExtracted pair_id range:")
print(f"Min: {features_df['pair_id'].min()}, Max: {features_df['pair_id'].max()}")

# Check overlap
print("\nOverlap check:")
feature_ids = set(features_df['pair_id'].unique())
pair_ids = set(pairs_df['id'].unique())
print(f"IDs in both: {len(feature_ids & pair_ids)}")
print(f"IDs only in features: {len(feature_ids - pair_ids)}")
print(f"IDs only in pairs: {len(pair_ids - feature_ids)}")

# Show some examples that don't match
print("\nSample non-matching feature IDs:")
print(sorted(feature_ids - pair_ids)[:20])
```
